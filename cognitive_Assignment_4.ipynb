```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

entry1 = {
    "question": "how do i change my account details",
    "answer": "Go to Settings > Account Details to update your information.",
    "keywords": "account details update change",
    "category": "account"
}

entry2 = {
    "question": "how can i check my fee payment status",
    "answer": "Go to Payments > Payment History to check your fee payment status.",
    "keywords": "payment fee status billing",
    "category": "billing"
}

fixed_entries.append(entry1)
fixed_entries.append(entry2)

df = pd.DataFrame(fixed_entries)
print(df)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        matches = query_words.intersection(keywords)
        if len(query_words) > 0:
            confidence = len(matches) / len(query_words)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })
    results.sort(key=lambda x: x["confidence"], reverse=True)
    return pd.DataFrame(results)


query = "how can i pay my fee"
result = score_query(query, df)

print(result)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]
category = "account"
result = same_category(category, df)
print(result)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        matches = query_words.intersection(keywords)

        if len(query_words) > 0:
            confidence = len(matches) / len(query_words)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })
    results.sort(key=lambda x: x["confidence"], reverse=True)
    if results:
        highest_score = results[0]["confidence"]
        results = [
            result for result in results
            if result["confidence"] == highest_score
        ]
    return pd.DataFrame(results)

query = "fee"
result = score_query(query, df)
print(result)
```
