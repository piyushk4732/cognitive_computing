## Q1. Basic operations on a 1-D array

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

## Q2(a). Reverse the NumPy array

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

## Q2(b)(i). Most frequent value and its indices

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

## Q2(b)(ii). Most frequent value and its indices

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

## Q3. Accessing elements using row and column indices

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

## Q4. linspace(), array properties and transpose

```python
import numpy as np

Piyush_Singh = np.linspace(10, 100, 25)

print("Array:", Piyush_Singh)
print("Dimensions:", Piyush_Singh.ndim)
print("Shape:", Piyush_Singh.shape)
print("Total elements:", Piyush_Singh.size)
print("Data type:", Piyush_Singh.dtype)
print("Total bytes:", Piyush_Singh.nbytes)

transposed = Piyush_Singh.reshape(25, 1)
print("Transpose using reshape():")
print(transposed)

print("Transpose using T:")
print(Piyush_Singh.T)
```

## Q5. 2-D array operations

```python
import numpy as np

ucs420_Piyush_Singh = np.array([[10, 20, 30, 40],
                                 [50, 60, 70, 80],
                                 [90, 15, 20, 35]])

print("Array:")
print(ucs420_Piyush_Singh)

print("Mean:", np.mean(ucs420_Piyush_Singh))
print("Median:", np.median(ucs420_Piyush_Singh))
print("Maximum:", np.max(ucs420_Piyush_Singh))
print("Minimum:", np.min(ucs420_Piyush_Singh))
print("Unique elements:", np.unique(ucs420_Piyush_Singh))

reshaped_ucs420_Piyush_Singh = ucs420_Piyush_Singh.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_Piyush_Singh)

resized_ucs420_Piyush_Singh = np.resize(ucs420_Piyush_Singh, (2, 3))
print("Resized array:")
print(resized_ucs420_Piyush_Singh)
```
