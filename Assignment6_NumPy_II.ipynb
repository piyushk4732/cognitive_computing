## Q1. Sensor readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

fahrenheit = (9 / 5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)

count = np.sum(corrected_temperature > 32)
print("Number of readings above 32°C:", count)

print("Vectorization applies the operation to the complete array at once, so there is no need to process each value separately with a loop. Boolean indexing also selects only the values that satisfy a condition, making the operation shorter and more efficient.")
```

## Q2. Daily steps

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum steps:", max_position)
```

## Q3. Array slicing, copying, reshaping and flattening

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("Using flatten():", flattened)
print("Using ravel():", raveled)

raveled[0] = 100
print("Matrix after modifying ravel():")
print(matrix)
print("Raveled array:", raveled)

flattened[1] = 200
print("Matrix after modifying flatten():")
print(matrix)
print("Flattened array:", flattened)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4. Cognitive-assistive system

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("X transpose:")
print(XT)

print("X.T @ X:")
print(XT @ X)

inverse = np.linalg.inv(XT @ X)
print("Inverse of X.T @ X:")
print(inverse)

beta = inverse @ XT @ y
print("OLS coefficients:", beta)

print("The three coefficients represent the effect of sleep hours, activity level, and stress level on the predicted assistance score when using the given linear model.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
